### $N_B$ for temporal change in allele frequency

#### 1. Summary
$N_B$ is the continuous version of Williamson and Slatkin (1999)'s full likelihood model on the changes in allele frequency between two or more time points. Continous here refers the treatment to allele count and $N_e$, while time remains as discrete generations. We know that binomial errors are assumed when sampling allele counts from true frequencies. And if we can use a beta distribution to model drift (i.e. change in true allele frqeuency over time), then beta and binomial work as a good pair (conjudate pair) in the modelling of the whole dynamics with both drift and sampling. This system of drift and sampling can also be viewed as a hidden Markov Chain, but let's no go there. 

It has a closed-form solution which is quite elegant. It is not only a solution to an equation but also a solution to some existing problems. Despite my code for $N_B$ was written in R, it was still 500x faster than Wang (2001) in Fortran. Imagine if somebody rewrites NB in C... Another improvement is that $N_B$ works for a wide range of $N_e$ from 50 to millions (virtually no upper bound). For the discrete version, the transimition matrix grows quadratially with $N_e$ hence it struggled when $N_e>20000$.

#### 3. Pseudo-replication among loci
$N_B$ assumes independent loci, which is generally not true. Linkage does not affect the mean allele frequency change for neutral loci, however, it reduces the amount of information the loci carry towards $N_e$. Hui, Brenas, Burt (2021) worked out the correlation among two changes in allele frequency over time and proposed a correction for pseudo-replication. 

#### 4. Combining samples from multiple time points 
Likelihood-based methods have the ability to combine drift signals across >2 time points. Temporal F struggles. See #5 below. 

#### 5. MAXTEMP (Waples et al. , 2025) and $N_B$ revisited
I had the pleasure of reviewing MAXTEMP, which claimed that adding a third temporal sample can refine an earlier temporal F (thus $N_e$) estimate. It is a seemingly outrangous claim, as the third sample is outside of the sampling horizon covered by the temporal F. In this system, there are two estimates for the temporal $N_e$ between the first two time points: the direct estimate, solely based on the drift between them, and an implied estimate, which makes use of the difference between the overall drift and that between the last two time points. A linear combination of the two gives the refined estiamte. 

In fact, all likelihood-based methods have been implementing this for >2 temporal samples behind the screen. In the $N_B$ paper, I fitted two different $N_e's$ to a three-sample scenario, then ran a LRT to test for constant $N_e$. Likelihood jointly consider all observations and all parameters simutaneously to find the best solutions during maximisation, while MAXTEMP is an iterative processs jumping back and forth in time. In short, the two share the same objective but with different ideologies, although personally I believe likelihood is more likely to return the optimal weighs. MAXTEMP's result approaches likelihood's asymptotically. 

On one hand, none of these was mentioned in $N_B$ and I blame the main author. On the other hand, it is probably quite subtle that it took the eagle-eyed Dr RS Waples ten years to figure it out. Nonetheless, I think it is nice for MAXTEMP to rediscover this phenomenon. I wrote a short article on this, see Hui (2025).
